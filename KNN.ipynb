```python
from sklearn.datasets import load_iris
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=load_iris() #entire dataset will be loaded in the the variable data
#splitting into feature and output classes
#do random splitting whilst training (not in knn)
```

```python
data
```

```python
X=data.data
Y=data.target
```

```python
#X[0] will be considered as datapoint or testing point
def euclidean_distance(X):
    dict1={}
    for i in range(1,150): #because X[0] is considered so we only need to go till 149 as 0-149 will be the accessing range
        dist=((X[0][0]-X[i][0])**2+(X[0][1]-X[i][1])**2+(X[0][2]-X[i][2])**2+(X[0][3]-X[i][3])**2)**0.5
        dict1[i]=dist
    return dict1
```

```python
result=euclidean_distance(X)
print(result)
```

```python
sorted_by_values=dict(sorted(result.items(),key=lambda item:item[1]))
```

```python
sorted_by_values
```

```python
k=10
i=0
for i in range(1,k+1):
    key_list=list(sorted_by_values)[:k]
print(key_list)
```

```python
#i=0 #no need for this 
target_list=[]
for i in key_list: #this gives us values of key_list i.e i in key_list gives us 17, 4, etc. and not indexes
    #target_list=[Y[i+1]] # this overwrites elements in every iteration
    target_list.append(Y[i+1]) #thereofre we need to append every element
print(target_list)
```

```python
unique, counts = np.unique(target_list, return_counts=True)
max_count = counts.max()
most_repeated = unique[counts == max_count]
print(most_repeated)
if len(most_repeated)!=1:
    print("Change the value of k as the element belongs to not just one class right now")
```
